```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
raw = pd.read_csv('../../data/raw/FM24.csv')
```

```python
num_cols = [
    'Acc', 'Agi', 'Bal', 'Jum', 'Nat.1', 'Pac', 'Sta', 'Str', 
    'Agg', 'Ant', 'Bra', 'Cmp', 'Cnt', 'Dec', 'Det', 
    'Fla', 'Ldr', 'OtB', 'Pos', 'Tea', 'Vis', 'Wor', 
    'Cor', 'Cro', 'Dri', 'Fin', 'Fir', 'Fre', 'Hea', 
    'Lon', 'L Th', 'Mar', 'Pas', 'Pen', 'Tck', 'Tec'
]
features = raw[num_cols].copy()
```

```python
meta_cols = [
    'Name', 'Club', 'Nat', 'Height', 'Weight', 'Age', 'CA', 'PA',
]
meta = raw[meta_cols].copy()
```

```python
meta.to_csv('../../data/processed/meta.csv', index=False)
features.to_csv('../../data/processed/raw_features.csv', index=False)
```

```python
feature_normalized = features.sub(features.mean(axis=1), axis=0).div(features.std(axis=1), axis=0)
feature_normalized.to_csv('../../data/processed/normalized.csv', index=False)
```
