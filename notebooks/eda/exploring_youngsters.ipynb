```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
raw = pd.read_csv('../../data/raw/FM24.csv')
```

```python
ca_above_120 = (raw['CA'] > 120).sum()
ca_below_120 = (raw['CA'] <= 120).sum()
age_above_18 = (raw['Age'] > 18).sum()

print(f"Players with CA > 120: {ca_above_120}")
print(f"Players with CA <= 120: {ca_below_120}")
print(f"Players with Age > 18: {age_above_18}")
```

```python
ca_above_120_and_age_below_18 = ((raw['CA'] > 120) & (raw['Age'] < 18)).sum()
print(f"Players with CA > 120 and Age < 18: {ca_above_120_and_age_below_18}")
```

```python
established_youngsters = raw[(raw['CA'] > 120) & (raw['Age'] < 18)]
print(established_youngsters[['Name', 'CA', 'Age']])
```
